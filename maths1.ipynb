```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
def predict_price(x):
    return 2 * x + 50

size = 1000
print(f"predicted price : ${predict_price(size)}")
```

```python
def relu(x):
    return np.maximum(0, x)
print(f"ReLU(-5):{relu(-5)} , ReLU(3):{relu(3)}")
```

```python
t = np.linspace(0,10,100)
accuracy = 1-1/(t + 1)

plt.plot(t,accuracy)
plt.title("increasing model accuracy")
plt.show()
```

```python
def training_process(x):
    return 2 * x if x < 5 else 0

print(f"4.9 : {training_process(4.9)} , At 5.0 : {training_process(5.0)}")
```

```python
def error_limit(n):
    return 5/n + 0.1

print(f"error with 1M samples : {error_limit(1000000)}")
```

```python
val = 0.5
for i in range(1,11):
    val *= 0.5
    print(f"step {i} : {val}")
```

```python
w = sp.Symbol('w')
l = w ** 2
derivative = sp.diff(l,w)
print(f"derivative at w=4 : {derivative.subs(w,4)}")
```

```python
t_vals = np.array([1,2,3])
d_vals = 10 - 2 * t_vals
slope = np.diff(d_vals) / np.diff(t_vals)
print(f"Rate of change : {slope[0]}")
```

```python
slopes = [0.5 , 0.01 , 0.0001]
for s in slopes:
    if abs(s) < 0.1:
        print(f"slope : {s} : close to converging")
```

```python
w_vals = np.linspace(-5,5,100)
loss = w_vals ** 2

print("sample weights (w) :",w_vals[::20])
print("sample loss (w^2) :",loss[::20])

second_derivative = 2
if second_derivative > 0:
    print("\n Result : The second derivative is positive (+2).")
    print("Shape : this is a 'valley' (concave up).")
    print("AI goal : Roll the ball to  the bottom (w = 0) to minimize loss!")
```

```python
def loss(w):
    return w ** 2 - 4 * w + 5

for w in [1,2,3]:
    print(f"weight {w} : {loss(w)}")
```

```python
w = 10
learning_rate = 0.1
slope = 2
w_new = w - learning_rate * slope
print(f"new weight : {w_new}")
```
