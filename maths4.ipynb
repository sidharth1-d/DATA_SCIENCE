```python
import pandas as pd
import numpy as np
from sympy.physics import qho_1d
import matplotlib.pyplot as plt
```

```python
data = [20,30,40,50,60,600]
df = pd.DataFrame(data,columns = ['watchTime'])

print(f"MEAN : {df['watchTime'].mean()}")
print(f"MEDIAN : {df['watchTime'].median()}")
print(f"MODE : {df['watchTime'].mode()[0]}")
```

```python
data = np.array([2,3,3,4,5,5,10])

data_range = np.max(data) - np.min(data)
q1,q3 = np.percentile(data,[25,75])
iqr = q3 - q1
print(f"Range : {data_range}")
print(f"Q1 (25th percentile) : {q1})")
print(f"Q3 (75th percentile) : {q3})")
print(f"IQR : {iqr})")

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 - 1.5 * iqr


outliers = data[(data < lower_bound) | (data > upper_bound)]

print(f"\n lower bound : {lower_bound}")
print(f" upper bound : {upper_bound}")
print(f"outliers : {outliers}")
```

```python
spending = np.array([100,200,300,400,500,600,700,800,900,1000])

threshold = np.percentile(spending , 90)

print(f"90th percentile threshold : ${threshold:.2f}")

top_users = spending[spending >= threshold]

print(f"top spenders : {top_users}")
```

```python
sensor_readings = np.array([50,51,49,50,50])

mean = np.mean(sensor_readings)
std_dev = np.std(sensor_readings)

print(f"mean distance : {mean:.2f}")
print(f"standard deviation : {std_dev:.2f}")

if std_dev < 1:
    print("sensor is stable")
else:
    print("sensor is noisy")
```

```python
data = np.array([10,12,13,15,100])

plt.boxplot(data , vert = False)
plt.title("box plot for outlier detection")
plt.xlabel("values")
plt.show()

q1,q3 = np.percentile(data , [25,75])
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = data[(data < lower_bound) | (data > upper_bound)]

print(f"Q1 : {q1}")
print(f"Q3 : {q3}")
print(f"IQR : {iqr}")
print(f"Lower bound : {lower_bound}")
print(f"upper bound : {upper_bound}")

print(f"Outerliers : {outliers}")
```

```python
ice_cream = np.array([10,20,30,40])
shark_attacks = np.array([1,3,5,8])

correlation = np.corrcoef(ice_cream,shark_attacks)[0,1]

print(f"correlation (r) : {correlation:.2f}")
```

```python
prices = np.array([100,120,130,150,160,500])

df_prices = pd.DataFrame(prices , columns = ['price'])

skewness = df_prices['price'].skew()
print(f"skewness : {skewness:.2f}")

plt.hist(prices  ,bins = 10 , density = True)

plt.title("Original price distribution (right skewed)")
plt.xlabel("price")
plt.ylabel("Density")

plt.show()

log_prices = np.log(prices)

plt.hist(log_prices , bins = 10 , density = True)

plt.title("log_transformed distributed (more Normal)")
plt.xlabel("Log(price)")
plt.ylabel("Density")

plt.show()

print(f"Skewness after log transform : {pd.Series(log_prices).skew():.2f}")
```

```python
income = np.array([40,50,60,70,80])

mean = np.mean(income)
std = np.std(income)

z_scores = (income - mean)/std

print(f"Mean : {mean}")
print(f"Standard deviation : {std:.2f}")
print(f"Z-scores : {z_scores}")
```

```python
machine_a = np.array([10,10,11,9])
machine_b = np.array([10,10,10,10])

mean_a = np.mean(machine_a)
mean_b = np.mean(machine_b)

var_a = np.var(machine_a)
var_b = np.var(machine_b)

print(f"machine A ---> Mean : {mean_a} , Variance : {var_a}")
print(f"machine B ---> Mean : {mean_b} , Variance : {var_b}")
```

```python
temps = pd.Series([20 , 21 ,22 ,20 , 0 , 21 ,500])

q1 = temps.quantile(0.25)
q3 =temps.quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Q1 : {q1} , Q3 : {q3} , IQR : {iqr}")
print(f"Lower bound : {lower_bound} , Upper bound : {upper_bound}")

filtered_temps = temps[
    (temps >= lower_bound) & (temps <= upper_bound)
]

print(f"\n Cleaned Data : {filtered_temps.tolist()}")
```
