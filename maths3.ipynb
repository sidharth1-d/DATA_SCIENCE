```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
total = 100
correct = 85
prob = correct / total
print(f"model accuracy (probability) : {prob:.2f}")
```

```python
p_tech = 0.3
p_finance = 0.2

p_either = p_tech + p_finance
p_both_tech = p_tech * p_tech

print(f"probability of tech or finance : {p_either}")
print(f"probability of two tech articles in a row : {p_both_tech}")
```

```python
p_spam = 0.1
p_offer_given_spam = 0.8
p_offer_given_safe = 0.05
p_safe = 0.9
p_offer = (p_offer_given_spam * p_spam) + (p_offer_given_safe * p_safe)
p_spam_given_offer = (p_offer_given_spam * p_spam) / p_offer

print(f"probability it's spam given the word 'offer' : {p_spam_given_offer:.2%}")
```

```python
np.random.seed(42)
session_time = np.random.exponential(scale = 2 , size = 1000)

plt.hist(session_time,bins = 30 , density = True)

plt.title("distribution of user session times")
plt.xlabel("minute")
plt.ylabel("density")
plt.show()
```

```python
mu = 5
sigma  = 0.5

np.random.seed(42)
speeds = np.random.normal(mu, sigma, 1000)

plt.hist(speeds,bins = 30 , density = True , alpha = 0.6)

xmin,xmax = plt.xlim()
x = np.linspace(xmin,xmax,1000)
p = norm.pdf(x,mu,sigma)

plt.plot(x,p)

lower = mu - sigma
upper = mu + sigma
plt.axvline(lower)
plt.axvline(upper)
plt.title("Robot speed distributionm (normal distribution)")
plt.xlabel("speed (KM/H)")
plt.ylabel("density")
plt.show()
```

```python
response_time = 250
mean = 200
std = 20

z_score = (response_time - mean)/std
print(f"Z-score : {z_score}")
```

```python
std_dev = 0.3
n = 100
se = std_dev / np.sqrt(n)
print(f"standard error : {se}")
```

```python
np.random.seed(42)
all_users = np.arange(1000)

night_users = np.random.choice(all_users , size = 150, replace = False)

biased_sample = np.random.choice(night_users , size = 50 , replace = False)

random_sample = np.random.choice(all_users , size = 50 , replace = False)

print("Biased sample (night users) : ")
print(biased_sample)

print("\n Random sample (All users) : ")
print(random_sample)
```

```python
np.random.seed(42)
true_ctr = 0.05
results = []
views_range = range(1,10000)

for n in views_range:

    clicks = np.random.binomial(1,true_ctr,n)
    ctr_estimate = np.mean(clicks)
    results.append(ctr_estimate)

plt.plot(views_range,results)
plt.axhline(true_ctr)

plt.title("law of larger numbers : CTR convergence")
plt.xlabel("number of Ad views")
plt.ylabel("estimated CTR")

plt.show()
```

```python
np.random.seed(42)

data = np.random.uniform(0,100,10000)

sample_size = 50
num_samples = 1000

sample_means = []

for _ in range(num_samples):
    sample = np.random.choice(data,sample_size)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)
plt.hist(sample_means , bins = 30 , density = True , alpha = 0.6)

mean = np.mean(sample_means)
std = np.std(sample_means)

x = np.linspace(min(sample_means),max(sample_means),100)
pdf = norm.pdf(x , mean , std)

plt.plot(x , pdf)

plt.title("central limit theorem : sample means distribution")
plt.xlabel("sample mean")
plt.ylabel("density")

plt.show()
```
