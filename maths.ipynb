```python
import numpy as np
```

```python
#data
weight = 200
bias = 50000
x = 1500

#linear regression

prediction = weight * x * bias
print(f"prediction: ${prediction}")
```

```python
def get_error(steps):
    '''
    simulates a linear error reduction model
    formula y=mx+c
    where m = -5 (slope) and c = 100 (starting error).
    '''
    return -5 * steps + 100

# testing the function with 10 steps

steps_input = 10
current_error = get_error(steps_input)

print(f"error after {steps_input} steps :{current_error}")

#eg---> seeing how error decreases over time

for s in range(0,21,5):
    print(f"at step {s:2},the error is : {get_error(s)}")
```

```python
user_a = np.array([2,3])
user_b = np.array([5,7])
distance = np.linalg.norm(user_a - user_b)
print(f"distance is : {distance}")
```

```python
king = np.array([1,0])
man = np.array([1,0])
apple = np.array([0,1])

#dot product (1 means same direction,0 means perpendicular)

print(f"king-man similarity : {np.dot(king, man)}")
print(f"king-apple similarity : {np.dot(king, apple)}")
```

```python
# a matrix that scales the x - direction by 5

a = np.array([[5,0],
              [0,1]])

#calculate

values , vectors = np.linalg.eig(a)
print(f"the multiplier (eigenvalues) is : {values[0]}")
print(f"the stable path (eigenvector) is : {vectors[:,0]}")
```

```python
# if the eigenvalue is 1, the data stays the same

current_viewer = np.array([100,100])
eigenvalue  = 1

next_day = eigenvalue * current_viewer
print(f"viewer tommorrow : {next_day}")
```
