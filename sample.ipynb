# **part -3**

```python
winner = df['winner'].value_counts()
winner.plot(kind = 'bar')
```

```python
venue = df['venue'].value_counts()
venue.plot(kind = 'pie',radius = 2,autopct = '%1.1f%%')
```

```python
toss = df['toss_decision'].value_counts(ascending = True)
toss.plot(kind = 'pie',radius = 1,autopct = '%1.1f%%')
```

# **part - 4**

```python
raw1 = df.groupby(['toss_decision', 'winner']).size()
raw1.plot(kind = 'bar',stacked = True)
```

# **part - 5**

```python
pivot = df.pivot_table(
    values='venue',
    index='toss_decision',
    columns='winner',
    aggfunc='count')

plt.figure(figsize=(14, 5))
sns.heatmap(
    pivot,
    annot=True
)
```
