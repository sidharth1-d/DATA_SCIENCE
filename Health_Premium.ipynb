```python
import  numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
age = np.array([18,25,35,45,55])
premium = np.array([15000 , 18000 , 21000 , 24000 , 29000])

def train_basic_regression(x,y):
    x_mean = np.mean(x)
    y_mean = np.mean(y)

    numerator = np.sum((x - x_mean)*(y - y_mean))
    denominator = np.sum((x - x_mean) ** 2)
    m = numerator / denominator

    b = y_mean - (m * x_mean)
    return m , b

slope,intercept = train_basic_regression(age , premium)

def predict_premium(user_age):
    return (slope * user_age) + intercept

test_age = 55
result = predict_premium(test_age)

print(f"--Health predictor (Basic)--")
print(f"cost increase per year (Slope): {slope:.2f}")
print(f"Base insurance per year (Intercept): {intercept:.2f}")
print(f"Base insurance premium for age {test_age}: {result:.2f}")
```

```python
data = {
    'income' : [20000 , 30000 , 50000 , 80000 , 15000, 25000 , 40000, 9000],
    'loan_amount' : [5000 , 10000 , 15000 , 20000 , 25000 , 30000 , 45000 , 10000],
    'default' : [0 ,0 ,0 ,0 , 1 , 1 , 1 , 0]
    }

df = pd.DataFrame(data)

x = df[['income' , 'loan_amount']]
y = df['default']

model = LogisticRegression()
model.fit(x,y)

new_applicant = pd.DataFrame([[30000 , 25000]],
                             columns=['income','loan_amount'])
prediction = model.predict(new_applicant)

probability = model.predict_proba(new_applicant)[0][1]

print(f"prediction : {'RISKY' if prediction == 1 else 'SAFE'}")
print(f"risk probability: {probability * 100:.2f}%")
```
